## One sample mean based T test ##

```python
import statistics

sample = [8.6, 9.4, 7.9, 6.8, 8.3, 7.3, 9.2, 9.6, 8.7, 11.4, 10.3, 5.4, 8.1, 5.5, 6.9] # Enter the sample
mean = statistics.stdev(sample)
print(f"The sample mean is {mean}")
```

```python
# @title Enter the mean, sample size, sample standard deviation, alpha, and choose the type of analysis
mean = 1.6722383060978339 # @param {"type":"number"}
sample_standard_deviation = 3 # @param {"type":"number"}
sample_size = 3 # @param {"type":"number"}
alpha = 0.05 # @param {"type":"number"}

import scipy.stats as stats
import numpy as np




# build the distribution mode
mu = mean #population mean, null hypothesis
s = sample_standard_deviation # sample standard deviation (estimated from population std for t-test)
n = sample_size # sample size
df = n - 1 # degrees of freedom for t-test

std_sample_mean = s / np.sqrt(n)
print(f"Population mean = {mu}\nSample std (s) = {s} \nSample size (n) = {n}\nDegrees of freedom (df) = {df}\nStandard error of sample mean = {std_sample_mean:.3f}\n")

# value of random variable from the sample
sample_mean = 16 # sample mean

# t_score for sample_mean in order to look up the table for probability
t_score = (sample_mean - mu) / std_sample_mean
print(f"The t_score for the sample mean random variable is {t_score:.3f}")

percentile = alpha
# critical value
t_alpha = stats.t.ppf(percentile, df) # Using t.ppf for a T-test


p_value = stats.norm.cdf(x = sample_mean, # value of the random variable (sample)
                         loc = mu,
                         scale = std_sample_mean
                         )
if (t_alpha < 0):
  print(f"The critical t-value for a {alpha} left-tailed test with {df} degrees of freedom is {t_alpha:.3f}")
  print(f"The p value for the significance level {alpha} with left-tailed test is {p_value}.")

elif (t_alpha > 0):
  print(f"The critical t-value for a {alpha} right-tailed test with {df} degrees of freedom is {t_alpha:.3f}")
  print(f"The p value for the significance level {alpha} with right-tailed test is {p_value}.")


if (p_value > alpha):
  print(f"Since the p value > {alpha} (statistically significant level), the sample data fails to reject the null hypothesis.")
elif (p_value < alpha):
  print(f"Since the p value < {alpha} (statistically insignificant level), the sample data rejects the null hypothesis.")
```
