## Proportions 1 Sample Z-Test ##

```python
# @title One Sample Z Test left-tail, right-tail, two-tail (Enter the sample data)
Population_mean = 5 # @param {"type":"number"}
alpha = 0.05 # @param {"type":"number"}
sample_success = 0.5 # @param {"type":"number"}
p_null = 0.05 # @param {"type":"number"}
Sample_size = 50 # @param {"type":"integer"}
tails = "two-tails" # @param ["one-tail","two-tails"]

import numpy as np
import scipy.stats as stats
import statistics
import plotly.graph_objects as go
from statsmodels.stats.proportion import proportions_ztest

sample = [8.6, 9.4, 7.9, 6.8, 8.3, 7.3, 9.2, 9.6, 8.7, 11.4, 10.3, 5.4, 8.1, 5.5, 6.9] # Enter the sample
n = Sample_size # sample size
alpha = alpha


# build the distribution mode
mu = Population_mean #population mean, null hypothesis


sample_mean = statistics.mean(sample)
std_sample = statistics.stdev(sample)
sigma = std_sample_mean * np.sqrt((n - 1) / n) # population std
print(f"Population mean = {mu: .3f}\nPopulation standard deviation = {sigma: .3f}\n\nSample mean = {sample_mean: .3f}\nSample mean standard deviation = {std_sample_mean: .3f}\n")


# z_score for sample_mean in order to look up the table for probability
z_score = (sample_mean - mu) / std_sample_mean
print(f"The z_score for the sample mean random variable is {z_score: .3f}")

p_value = stats.norm.cdf(x = sample_mean,
                          loc = mu,
                          scale = std_sample_mean
                          )

# critical value
z_alpha = (stats.norm.ppf(alpha)) * -1

if (tails == "one-tail"):
  if (z_score < 0):
    print(f"The p value for the significance level {alpha} with left-tailed test is {p_value}")
    print(f"The alpha (critical value) for {alpha} left-tailed test is {z_alpha: .3f}")
  elif (z_score > 0):
    print(f"The p value for the significance level {alpha} with right-tailed test is {p_value}")
    print(f"The alpha (critical value) for {alpha} right-tailed test is {z_alpha: .3f}")



# two sided sample z test
elif(tails == "two-tails"):
  #output z_stat and p value
  type_test = "two-sided"
  z_stat, p_value = proportions_ztest(
                                      count = sample_success, #number of successes in the sample
                                      nobs = n,#number of observations
                                      value = p_null,
                                      alternative = type_test,
                                      prop_var = False # False means to use sample data for variance

                                    )

  print(f"The z test stat = {z_stat: .3f} and p value = {p_value}")

if (p_value > alpha):
  print(f"Since the p value > {alpha} (statistically significant level), the sample data fails to reject the null hypothesis.")
elif (p_value < alpha):
  print(f"Since the p value < {alpha} (statistically insignificant level), the sample data rejects the null hypothesis.")
```

```python
# @title 預設標題文字
A_success = 0 # @param {"type":"number"}
B_success = 0 # @param {"type":"number"}
A_size = 0 # @param {"type":"integer"}
B_size = 0 # @param {"type":"integer"}
test_type = "two-sided" # @param ["two-sided","larger","smaller"]

from statsmodels.stats.proportion import proportions_ztest
import numpy as np

# Assign values to number of successes and sample sizes for both samples

sample_success_A, sample_size_A = (A_success, A_size)
sample_success_B, sample_size_B = (B_success, B_size)

sample_sucess = np.array([sample_success_A, sample_success_B])
sample_size = np.array([sample_size_A, sample_size_B])

if (test_type == "two-sided"):
  test_type = "two-sided"
elif (test_type == "smaller"):
  test_type = "smaller"
elif(test_type == "larger"):
  test_type = "larger"


z_stat, p_value = proportions_ztest(
                                    count = sample_sucess,
                                    nobs = sample_size,
                                    alternative = test_type
                                    )


print(f"z-stat = {z_stat: .4f}, p-value = {p_value}")

```
