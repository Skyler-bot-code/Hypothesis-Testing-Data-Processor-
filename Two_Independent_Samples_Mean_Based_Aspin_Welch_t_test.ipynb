The function for the standard error and degree of freedom for the Aspin-Welch t-test

```python
import numpy as np

def se_df(n1, s1, n2, s2):
  """
  se: standard error
  df: degrees of freedom
  """
  se = np.sqrt(s1 ** 2 / n1 + s2 ** 2 / n2)
  den_1 = (s1 ** 2 / n1) ** 2 / (n1 - 1)
  den_2 = (s2 ** 2 / n2) ** 2 / (n2 - 1)

  df = se ** 4 / (den_1 + den_2)
  return se, df
```

```python
import scipy.stats as stats
import numpy as np

# Sample 1
sample1_o = [
            67.6, 41.2, 85.3, 55.9, 82.4, 91.2, 73.5, 94.1, 64.7, 64.7, 70.6,\
            38.2, 61.8, 88.2, 70.6, 58.8, 91.2, 73.5, 82.4, 35.5,94.1, 88.2, 64.7,\
            55.9, 88.2, 97.1, 85.3, 61.8, 79.4, 79.4
            ]

# Sample 2
sample2_f = [
             77.9, 95.3, 81.2, 74.1, 98.8, 88.2, 85.9, 92.9, 87.1, 88.2, 69.4, 57.6,\
             69.4, 67.1, 97.6, 85.9, 88.2, 91.8, 78.8, 71.8, 98.8, 61.2, 92.9, 90.6,\
             97.6, 100, 95.3, 83.5, 92.9, 89.4
             ]

mean_o = np.mean(sample1_o) #independent group 1
n1 = np.size(sample1_o) # sample size of group 1
s1 = np.std(sample1_o) # sample standard deviation for group 1

mean_f = np.mean(sample2_f) #independent grou 2
n2 = np.size(sample2_f) # sample size of group 2
s2 = np.std(sample2_f) # sample standard deviation for group 2

# check to see which side of the mu = 0 the sample statistic is (mean_o - mean_f)
mean_stat = mean_o - mean_f

# compute the parameters for the hypothesis test distribution

se, df = se_df(n1, s1, n2, s2)
mu = 0 # null hypothesis
p_value = stats.t.cdf(mean_stat,
                      df = df,
                      loc = mu,
                      scale = se
                      )

print(f"The p value for the test = {p_value}")
print(f"Online course: means = {mean_o}, sample size = {n1}, and standard deviation = {s1}")
print(f"face-to-face course: means = {mean_f}, sample size = {n2}, and standard deviation = {s2}")
```

Direct way to find the Aspin-Welch t-test using scipy

```python
import scipy.stats as stats

# @markdown ## Select the type of alternative hypothesis:
alternative_test_type = 'greater' # @param ['less', 'greater', 'two-sided']


stats.ttest_ind(sample1_o, sample2_f, equal_var = False, alternative = alternative_test_type)
```

Direct way to find the Aspin-Welch t-test using statsmodels

```python
import statsmodels.stats.weightstats as sm
# @markdown ## Select the type of alternative hypothesis:
alternative_test_type = 'two-sided' # @param ['smaller', 'larger', 'two-sided']

# set usevar = "unequal" to use Aspin Weltch t test

t_stat, p_value, df = sm.ttest_ind(sample1_o, sample2_f, alternative = alternative_test_type, usevar = "unequal")

print(f"The t-statistic is {t_stat}, p-value = {p_value}, and df = {df}")
```
