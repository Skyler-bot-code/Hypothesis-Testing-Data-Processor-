## Chi-Square Test of Independence ##



```python
# Enter the data

data =

observe = pd.DataFrame(data)
```

**Compute the Expected Value**

Find the column total

```python
col_total = [sum(observe["Column 1"]), sum(observe["Column 2"]), sum(observe["Column 3"]), sum(observe["Column 4"]), sum(observe["Column 5"])]
```

Find the row total

```python
row_total = observe.sum(axis = 1) # col totals axis = 0 sum of each column
row_total
```

```python
import numpy as np

expected = np.outer(row_total, col_total) / np.sum(row_total)
expected
```

Expected frquencies DataFrame

```python
expected = pd.DataFrame(expected,
                        index = observe.index,
                        columns = observe.columns
                        )

expected
```

Find the chi-square statistic, degree of freedom, and p-value

```python
import scipy.stats as stats

# the chi-square statistic is always positive
chi_stat = ((observe - expected) ** 2 / expected).sum().sum() # sum of all the rows first and then sum of all the columns

print(f"Chi-square statistic = {chi_stat: .4f}")

# df = degree of freedom
df = (observe.shape[0] - 1) * (observe.shape[1] - 1) # table.shape or matrix.shape will give row and column numbers

print(f"The degree of freedom = {df}")

# Find the p value
p_value = 1 - stats.chi2.cdf(
                            x = chi_stat,
                            df = df
                            )

print(f"The p value = {p_value}")
```

## Chi-Square Goodness of Fit ##

Find the expected frequency

```python
import numpy as np
import scipy.stats as stats

# create two arrays in numpy
# only numpy array can perform arthemetic operations and native python list type cannot

observed = np.array([]) # array of the percentage in observed

expected = np.array([]) / 100 # concert the number to decimal values
expected = expected * observed.sum() # convert the percent into counts of the total observed


print(f"The expected equally frequnecy distribution = {expected}")
```

Compute the statistics

```python
chi_stat = ((observed - expected) ** 2 / expected).sum()
df = len(expected) - 1
print(f"The chi square statistic = {chi_stat: .4f}")
print(f"The degree of freedom = {df}")

# compute the p value (right-tailed)

p_value = 1 - stats.chi2.cdf(x = chi_stat, df = df)
print(f"p value = {p_value}")
```
